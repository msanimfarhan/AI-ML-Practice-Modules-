# Uni-Variate Analysis (Categorical valuses so countplot)

```python
sns.countplot(data=df,x=df['Survived'])
```

```python
(df['Survived'].value_counts() / len(df)) * 100 
```

```python
survived_count = df['Survived'].value_counts() 
survived_label = df['Survived'].unique()
print(survived_label)
plt.pie(survived_count,labels=survived_label, autopct="%1.1f")
plt.title('Pie chart of survival rate')
plt.show()
```

```python
sns.countplot(data=df,x=df['Pclass'])
```

```python
(df['Pclass'].value_counts() / len(df) ) * 100
```

```python
survived_count = df['Pclass'].value_counts()
print(survived_count)
survived_label = df['Pclass'].unique()
print(survived_label)
```

```python
plt.pie(survived_count,labels=survived_label,autopct="%1.1f")
```

# uni-variate analysis for nuemerical values

```python
sns.histplot(df['Age'],bins=50)
plt.title('Histogram of Age')
plt.xlabel('Age')
plt.ylabel('Frequency') 
plt.show()
```

```python
sns.kdeplot(df['Age']) 
plt.title('Kde plot of Age')
plt.xlabel('Age')
plt.ylabel('Probability Distribution') 
plt.show()
```

```python
sns.histplot(df['Fare'],bins=50)
plt.title('Histogram of Fare')
plt.xlabel('Fare')
plt.ylabel('Frequency') 
plt.show()
```

```python
sns.kdeplot(df['Fare']) 
plt.title('Kde plot of Fare')
plt.xlabel('Fare')
plt.ylabel('Probability Distribution') 
plt.show()
```
