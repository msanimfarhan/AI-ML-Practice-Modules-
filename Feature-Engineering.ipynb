```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns 
from sklearn.model_selection import train_test_split
```

```python
df.drop(['PassengerId','Name','Ticket'],axis=1,inplace=True)
```

```python
# missing values
df.isnull().sum()
```
